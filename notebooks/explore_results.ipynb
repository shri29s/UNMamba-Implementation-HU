```python
import matplotlib.pyplot as plt
```

```python
plt.plot(history["epochs"], history["losses"])
plt.title("Loss vs Epoch")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.show()
```

```python
for i, map in enumerate(history["M_history"][-1:]):
    map = map.squeeze()
    plt.figure(figsize=(9, 9))
    
    for i in range(len(map)):
        plt.subplot(2, 2, i + 1)
        plt.imshow(map[i], cmap="jet")
    
    plt.savefig(f"../results/M{i}.png")
```
